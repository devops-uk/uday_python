## Taking help: Keywords, Identifiers and modules...etc:

In Python, **keywords** are predefined and reserved words with specific meanings. The syntax of the coding is defined by keywords. The keyword is not allowed to be used as an identifier, function name, or variable name. Except for True and False, all keywords in Python are written in lower case. Python 3.7 contains 33 keywords; let's go over them all one by one.

An **identifier** is a name that is used to identify a variable, function, class, module, and so on. The identifier is a string of characters followed by an underscore. The identifier should begin with a character or an Underscore and end with a digit. The letters are A-Z or a-Z, an underscore (_), and digits (0-9). Special characters (#, @, $,%,!) should not be used in identifiers.

**Mutable and immutable:**  
dictionaries (dict) in Python are mutable, meaning you can modify their contents after they have been created. This includes adding, removing, or updating key-value pairs within the dictionary. Here's an example demonstrating the mutability of dictionaries:

````python
# Create a dictionary
my_dict = {'a': 1, 'b': 2, 'c': 3}
print(my_dict)
# Modify the dictionary
my_dict['d'] = 4  # Adding a new key-value pair
del my_dict['a']  # Removing a key-value pair
my_dict['b'] = 20  # Updating the value associated with a key
print(my_dict)  # Output: {'b': 20, 'c': 3, 'd': 4}
````

In this example, you can see that we added a new key-value pair ('d': 4), removed the key 'a' and its associated value, and updated the value associated with the key 'b'. These operations demonstrate the mutability of dictionaries in Python.


In Python, **the keys of a dictionary are immutable**, which means they cannot be modified after they have been created. **Once a key-value pair is added to a dictionary, the key cannot be changed**.

However, you can achieve a similar effect by creating a new key with the desired value and then deleting the old key. Here's an example:

````python
# Create a dictionary
my_dict = {'a': 1, 'b': 2, 'c': 3}

# Modify a key by creating a new key and deleting the old one
my_dict['new_key'] = my_dict.pop('a')

print(my_dict)  # Output: {'b': 2, 'c': 3, 'new_key': 1}
````

In this example, we create a new key 'new_key' with the value that was associated with the old key 'a', and then we remove the old key 'a' using the pop() method. This effectively "modifies" the key in the dictionary.

In Python, **bytes** and **bytearray** are built-in data types used to represent sequences of bytes.

**bytes** is an **immutable sequence type**, which means that once it is created, it cannot be changed. **bytearray, on the other hand, is a mutable sequence type**, which allows you to modify the contents of the sequence after it has been created.

````python
# create a bytes object
b = b"hello world"

# create a bytearray object
ba = bytearray(b)
print(ba)
# modify the bytearray
ba[0] = ord('H')

# convert the modified bytearray back to bytes
b_new = bytes(ba)

# print the new bytes object
print(b_new)
````

This code creates a bytes object b with the value "hello world", and then creates a bytearray object ba from the bytes object. The code then modifies the first byte of the bytearray to be the ASCII code for the letter "H". Finally, the modified bytearray is converted back to a bytes object and printed to the console. The output should be b'Hello world'.

In summary, **bytes and bytearray are useful for working with binary data in Python**, such as **reading and writing binary files or working with network protocols**. bytes is immutable, while bytearray is mutable  
**bytes** and **bytearray** are useful when working with binary data, such as reading or writing binary files, network protocols, cryptography, and encoding and decoding data.

````python
# put bags into the queue
q.put('bag1')
print(q.empty())
q.put('bag2')
q.put('bag3')
````

````python
# get bags from the queue in FIFO order
print(q.get())
print(q.get())
print(q.get())
````
