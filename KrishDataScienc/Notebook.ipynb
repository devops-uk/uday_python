## Setting up your machine (step by step Installation on windows)

- You can download Python **directly from its project site:** ([https://www.python.org/downloads/](https://www.python.org/downloads/)) and install individual components and libraries you want
- Alternately, you can download and install a package using **Anaconda**, which comes with pre-installed libraries (recommended).
